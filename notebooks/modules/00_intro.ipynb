# Databricks Fundamental
**1-Day Training | RetailHub Scenario**

Welcome to the training! Today we go from zero to a working, scheduled data pipeline in Databricks:
the platform and its costs, loading and exploring data, Delta Lake, jobs and permissions.

**Target audience:** Data analysts, BI specialists, data engineers and technical staff starting to work with Databricks.

**Prerequisites:** Basic SQL. No Python experience needed: every PySpark operation is shown next to its SQL equivalent.

**Scenario:** Throughout the day we work with a fictional company **RetailHub**, an online clothing store with customers,
orders and products. The data flows through the Bronze → Silver → Gold (medallion) architecture.

## Learning Objectives

After completing the training you will be able to:

- **Navigate** the workspace: notebooks, Git folders, Catalog Explorer, SQL Editor
- **Choose** compute (Serverless, classic cluster, SQL warehouse) and **control costs**: job vs all-purpose compute, autoscaling, auto-termination
- **Use** Unity Catalog's three-level namespace (`catalog.schema.table`) and Volumes for files
- **Explain** how Spark runs your code: driver and executors, lazy evaluation, one engine for SQL and PySpark
- **Load** CSV and JSON with an explicit schema and **write** the result to a Delta table (medallion architecture)
- **Explore** data with `display()`, `show()`, `describe()`, `summary()`, Edit Chart and Data Profile
- **Transform** data with `select`, `filter`, `withColumn`, `cast` / `try_cast`, `when`, `groupBy`, each with its SQL equivalent
- **Ask Genie Code** (formerly Databricks Assistant) to explain, write and fix code
- **Change and recover** Delta tables: INSERT, UPDATE, DELETE, MERGE, history, time travel, RESTORE
- **Build** a Lakeflow Job (formerly Workflows) with tasks, parameters, a schedule and a repair run
- **Grant** access with account groups: `GRANT`, `SHOW GRANTS`, `REVOKE`
- **See** how a Genie Agent answers natural-language questions on Gold data (optional trainer demo)

## Getting the training materials — Git folder

All notebooks and data come from one public Git repository. You clone it once into your workspace as a **Git folder**,
and pull updates whenever the trainer publishes a fix.

**Clone (once)**
1. Left menu → **Workspace** → **Home** (your user folder)
2. **Create** → **Git folder**
3. Git repository URL: `https://github.com/ditw-training/databricks-fundamental` — provider **GitHub** is detected automatically;
   the repository is public, so no credentials are needed
4. **Create Git folder** → the folder `databricks-fundamental` appears in your Home
5. Open `notebooks/modules/00_intro` from that folder

**Pull (get updates)**
1. Open the **Git dialog**: in the Workspace browser click **Git** beside the folder name — or, in any notebook from the
   folder, click the branch button (`main`) next to the notebook name
2. Click **Pull** → notebooks are updated to the latest commit
3. Conflict? It means you edited a notebook the trainer also changed. Copy your notebook outside the Git folder
   (right-click → **Clone**), then in the Git dialog use **⋮ → Reset** to discard your changes, and **Pull** again

**Commit & Push (your own repositories)**
1. Git dialog → changed files are listed on the left → tick the files
2. Enter a commit message → **Commit & Push**
3. Push needs write access to the remote — in this training you only **Pull** from the course repository

> The same dialog lets you create and switch **branches**. More on Git folders in M01.
