# Task 1: Validate Source

Validates row count against `min_rows` threshold. Returns status + count via `dbutils.notebook.exit()`.

> Same code as in `notebooks/modules/05_orchestration_jobs.ipynb`. `dbutils.notebook.exit()` stays in the last cell on purpose — anything printed in that cell would be hidden in the job run output.


```python
# TASK 1: Validate Source Data

# Parameters from Job
dbutils.widgets.text("source_table", "samples.nyctaxi.trips")
dbutils.widgets.text("min_rows", "100")

source_table = dbutils.widgets.get("source_table")
min_rows = int(dbutils.widgets.get("min_rows"))

# Validation
df = spark.table(source_table)
row_count = df.count()

if row_count < min_rows:
    raise Exception(f"Validation FAILED: {row_count} rows < {min_rows} minimum")

# Pass the row count to downstream tasks as a task value
dbutils.jobs.taskValues.set(key="row_count", value=row_count)
```

```python
# Exit value is visible in the task run output (not readable by other tasks)
import json
dbutils.notebook.exit(json.dumps({
    "status": "SUCCESS",
    "source_table": source_table,
    "row_count": row_count
}))
```
